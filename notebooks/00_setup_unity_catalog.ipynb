```python
%sql
-- Catálogo del proyecto
CREATE CATALOG IF NOT EXISTS dunnhumby
COMMENT 'Proyecto final Fundamentos de Databricks: Dunnhumby The Complete Journey';

USE CATALOG dunnhumby;

-- Un esquema por capa Medallion
CREATE SCHEMA IF NOT EXISTS bronze COMMENT 'Datos crudos tal como llegan de Kaggle';
CREATE SCHEMA IF NOT EXISTS silver COMMENT 'Datos limpios y con tipos corregidos';
CREATE SCHEMA IF NOT EXISTS gold   COMMENT 'Tablas agregadas para las preguntas de negocio';

SHOW SCHEMAS;
```

```python
%sql
CREATE VOLUME IF NOT EXISTS dunnhumby.bronze.raw_files
COMMENT 'CSV originales de Kaggle sin modificar';
```
